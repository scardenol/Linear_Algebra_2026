# Ejemplo para la matriz $A_{2\times 2}$ dada

## Parte 1

```python
# Importar librerías
import numpy as np
from scipy import linalg
```

```python
# Parámetros
A = np.array([[3,2],[2,3]])
xi = np.array([[1],[2]])
k=8
```

```python
# Ciclo principal
for _ in range(k):
    norm=np.linalg.norm(xi)
    yi = (1/norm)*xi
    xi = A @ yi
    Ri = yi.T @ xi #Transpuesta de yi por xi
    print(f"Iteración={_}, yi={np.round(yi, 6).tolist()}, Ri={Ri[0][0]:.6}\n")
```

```python
#Para validar el resultado
valores, vectores = linalg.eig(A)
print(f"\nvalores={np.round(valores, 6).tolist()}, \nvectores={np.round(vectores, 6).tolist()}")
```

## Parte 2

```python
# Parámetros
l1 = Ri # lambda_1 = R_k
B = A - l1*np.identity(2)
xi = np.array([[1],[2]])
k=4
```

```python
# Ciclo principal
for _ in range(k):
    norm=np.linalg.norm(xi)
    yi = (1/norm)*xi
    xi = B @ yi
    Ri = yi.T @ xi #Transpuesta de yi por xi
    print(f"Iteración={_}, yi={np.round(yi, 6).tolist()}, Ri={Ri[0][0]:.6}\n")
```

```python
#Para validar el resultado
valores, vectores = linalg.eig(B)
print(f"\nvalores={np.round(valores, 6).tolist()}, \nvectores={np.round(vectores, 6).tolist()}")
```

```python
# Valor propio de A lambda_r más alejado del dominante lambda_1
lr = l1 + Ri
print(f"lambda_k={lr[0][0]:.6}")
```
